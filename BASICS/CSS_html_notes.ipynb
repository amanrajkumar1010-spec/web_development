## Basic of css and html


Here are **short notes** for your HTML code:

### 📝 HTML Short Notes

* `<html>` → Defines the beginning and end of an HTML document.
* `<body>` → Contains the visible content of the webpage.
* `<h1>` → Creates the **largest heading**.
* `art1="Hero"` → A custom attribute; normally, `id="Hero"` or `class="Hero"` would be used instead.
* `<p>` → Defines a paragraph.
* `<br>` → Adds a **line break**.
* `<b>` → Makes text **bold**.
* `<img>` → Displays an image.
* `src` → Specifies the **image URL or file path**.
* `width="200px"` → Sets the image width to 200 pixels.
* `alt` → Provides **alternative text** if the image cannot be displayed and helps accessibility.
* `Spider.jpg` → Refers to an image file named `Spider.jpg` in the same folder as the HTML file.
* `<a>` → Creates a hyperlink.
* `href` → Specifies where the link goes.
* `target="_blank"` → Opens the link in a **new tab**.
* `</body>` and `</html>` → Close the body and HTML document.
* !+Enter ==> Make a full Html structure (boiler color code of html)
HTML Short Notes
* `<!DOCTYPE html>` → Tells the browser that the document uses HTML5.
* `<html lang="en">` → The root element of the HTML page; lang="en" specifies English as the language.
* `<head>` → Contains information about the webpage that is generally not displayed directly on the page.
* `<meta charset="UTF-8">` → Sets the character encoding to UTF-8, allowing many different characters and symbols to display correctly.
* `<meta name="viewport" ...>` → Makes the webpage responsive on different screen sizes, especially mobile devices.

**Important:** For an image, `src` should contain the actual image URL/file, while `alt` should contain a description such as `"Spider-Man"`.


## 1. HTML Structure & Concepts

* **`<div>`** : Elements ko group karne ya container banane ke liye use hota hai.
* **Parent Element** : Jis element ke andar doosre elements hote hain.
* **Child Element** : Jo element kisi parent container ke andar rehta hai.
* **`<main>`** : Webpage ke primary/main content ko represent karta hai.
* **`id`** : Kisi element ko unique naam/pehchaan dene ke liye use hoti hai; CSS mein ise `#id_name` se target kiya jata hai.

---

## 2. CSS Flexbox Basics

* **`display: flex;`** : Parent container par lagaya jata hai taaki uske direct children ka layout control ho sake.
* **Flex Direction (Default)** : By default flex ki direction **`row`** hoti hai.
  * **`row`** : Children side-by-side (ek line mein aamne-saamne) align hote hain.
  * **`column`** : Children upar se niche (vertical stack) mein align hote hain.

---

## 3. Important Layout Rule

> **Dhyan de:** Agar aap `<main>` par `display: flex;` lagate hain, toh sirf `<main>` ke direct children hi flex layout ka hissa honge. `<main>` ke bahar wala element (jaise `Box3`) us flex context ka part **nahi** hota.
